```python
import pandas as pd
import random
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

spam_samples = [
    "Win a lottery now", "Cheap medicines available", "Claim your prize today",
    "Get rich quick scheme", "Congratulations you won"
]
ham_samples = [
    "Meeting tomorrow at 10 AM", "Project deadline next week", "Family dinner tonight",
    "Work report submission", "Team discussion scheduled"
]

data = []
for i in range(50):  # 50 spam
    data.append({"Email": random.choice(spam_samples), "Label": "Spam"})
for i in range(50):  # 50 ham
    data.append({"Email": random.choice(ham_samples), "Label": "Ham"})

df = pd.DataFrame(data)

X = df["Email"]
y = df["Label"]

vectorizer = CountVectorizer(stop_words="english")
X_counts = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_counts, y, test_size=0.3, random_state=42)

model = MultinomialNB()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```

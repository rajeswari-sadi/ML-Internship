```python
import pandas as pd
import random
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report

# Synthetic dataset
genres = ["Sci-Fi", "Thriller", "Romance", "Action", "Comedy"]
plots = {
    "Sci-Fi": ["Spaceship fights aliens", "Humans colonize Mars", "Robot uprising", "Time travel adventure"],
    "Thriller": ["Detective solves murder", "Secret agent uncovers conspiracy", "Kidnapping mystery", "Psychological suspense"],
    "Romance": ["Couple falls in love", "Romantic Paris trip", "Love triangle drama", "Wedding story"],
    "Action": ["Superhero saves city", "Car chase explosion", "Martial arts battle", "Rescue mission"],
    "Comedy": ["Friends go on adventure", "Funny family vacation", "Comedy of errors", "Stand-up show"]
}

data = []
for genre in genres:
    for i in range(20):  # 20 samples per genre
        data.append({"Plot": random.choice(plots[genre]), "Genre": genre})

df = pd.DataFrame(data)

X = df["Plot"]
y = df["Genre"]

vectorizer = TfidfVectorizer(stop_words="english")
X_tfidf = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_tfidf, y, test_size=0.3, random_state=42)

model = MultinomialNB()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```
